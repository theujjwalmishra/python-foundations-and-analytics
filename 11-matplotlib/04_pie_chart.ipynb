```python
import matplotlib.pyplot as plt
```

```python
players = ["Sachin","Sehwag","Kohli","Rohit"]
color = ["Red","blue","green","yellow"]
runs = [123+345+342+431,553,973,652]
plt.pie(runs,labels=players,colors=color)
plt.title("Runs Comparision")
plt.show()
```
