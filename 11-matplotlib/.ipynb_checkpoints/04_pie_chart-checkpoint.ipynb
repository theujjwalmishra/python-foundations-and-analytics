```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli", "Rohit"]
runs = [123 + 345 + 342 + 431, 553, 973, 652]  # Sachin = 1241
explode = [0.2, 0, 0, 0]

plt.pie(
    runs,
    labels=players,
    colors=["Red", "blue", "green", "yellow"],
    explode=explode,
    shadow=True,
    startangle=90,
    autopct="%1.1f%%",  # Correct percentage formatting
    pctdistance=0.8,  # Percentage values ko thoda baahar push karne ke liye
    labeldistance=0.4,  # Player names ko chart ke ANDAR lane ke liye (default 1.1 hota hai)
    textprops={
        "color": "black",
        "weight": "bold",
    },  # Indar text saaf dikhne ke liye white aur bold
)

plt.title("Runs Comparison")
plt.show()
```

```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli", "Rohit"]
runs = [123 + 345 + 342 + 431, 553, 973, 652]
explode = [0.2, 0, 0, 0]

# Pre-calculate totals to match the labels by index
labels_iter = iter(players)

plt.pie(
    runs,
    colors=["Red", "blue", "green", "yellow"],
    explode=explode,
    shadow=True,
    startangle=90,
    autopct=lambda pct: next(labels_iter),  # Percentage ki jagah direct name print karega
    pctdistance=0.5,  # Names ko exact slice ke beech mein rakhne ke liye
    textprops={"color": "Black", "weight": "bold", "fontsize": 11},
)

plt.title("Runs Comparison")
plt.show()
```
