```python
import matplotlib.pylab as plt
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]
```

```python
plt.scatter(study_hours,exam_scores , color="red")
plt.xlabel(study_hours)
plt.ylabel(exam_scores)
plt.title("Student")
plt.show()
```
