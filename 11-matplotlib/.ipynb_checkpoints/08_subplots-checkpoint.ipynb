```python
import matplotlib.pylab as plt
```

```python
# 1 X 2 Grid
```

```python
year = [2015,2016,2017,2018]
rohit = [432,543,654,876]
virat = [654,765,876,762]
plt.subplot(1,2,1)
plt.plot(year,rohit,label="Rohit",color="Blue")
plt.title("Rohit")
plt.legend()

plt.subplot(1,2,2)
plt.plot(year,virat,label="Virat",color="red")
plt.title("Virat")
plt.legend()
plt.show()
```

```python
# 2 X 2 Grid
```

```python
sachin = [567,678,789,896]
sehwag = [3345,456,678,765]
plt.subplot(2,2,1)
plt.plot(year,sachin,label="Sachin")
plt.title("Sachin")
plt.subplot(2,2,2)
plt.plot(year,sehwag,label="Sehwag")
plt.title("Sehwag")
plt.subplot(2,2,3)
plt.plot(year,rohit,label="Rohit",color="Blue")
plt.title("Rohit")


plt.subplot(2,2,4)
plt.plot(year,virat,label="Virat",color="red")
plt.title("Virat")

plt.legend()
plt.tight_layout()
plt.show()
```
