```python
import pandas as pd
```

```python
data = {
    "Name": ["Alice","Bob","Harry","Ujjwal"],
    "Maths": [90,84,73,89],
    "Science": [90, 82, 89,98],
    "English": [88, 85, 94,76]
}
df = pd.DataFrame(data)
df
```

```python
df_2 = df.melt(id_vars=["Name"],value_vars=["Maths","Science","English"],var_name="Subject",value_name="Marks")
df_2
```

```python
df_2.to_csv("Pivot.csv" , index=False) # used to create a new csv
```
